## Your final prompt

> Paste your complete, finished prompt below (double-click to edit).

```
Write a complete Python program for beginners that acts as a Budget Purchase Advisor.

The program should:

1. Display a welcome message.
2. Ask the user for:
   - Their monthly income (float)
   - Their monthly expenses (float)
   - The price of the item they want to buy (float)
3. Calculate:
   - Remaining money = monthly income - monthly expenses
   - The percentage of the remaining money that the purchase would use.
4. Use an if-elif-else statement to make a decision:
   - If the purchase price is greater than the remaining money, display "You cannot afford this purchase."
   - Else if the purchase uses more than 50% of the remaining money, display "Think carefully before buying. This is a large expense."
   - Else if the purchase uses between 20% and 50% of the remaining money (inclusive), display "This purchase is reasonable, but make sure it fits your budget."
   - Otherwise, display "This purchase fits comfortably within your budget."
5. Display:
   - Monthly income
   - Monthly expenses
   - Remaining money
   - Purchase price
   - Percentage of remaining money used by the purchase (rounded to one decimal place)
   - The recommendation message.
6. If the remaining money is less than or equal to zero, display "Your expenses are greater than or equal to your income. Focus on reducing expenses before making purchases." Do not calculate the percentage in this case.
7. Use clear variable names, comments explaining each section, and only basic Python features suitable for beginners (input(), float(), print(), arithmetic, if-elif-else, and f-strings).
8. Produce the complete program only, with no explanations outside the code.
```

````python
# Budget Purchase Advisor

# Welcome message
print("=== Budget Purchase Advisor ===")

# Get user input
monthly_income = float(input("Enter your monthly income: $"))
monthly_expenses = float(input("Enter your monthly expenses: $"))
purchase_price = float(input("Enter the price of the item: $"))

# Calculate remaining money
remaining_money = monthly_income - monthly_expenses

print("\n----- Budget Summary -----")
print(f"Monthly income: ${monthly_income:.2f}")
print(f"Monthly expenses: ${monthly_expenses:.2f}")
print(f"Remaining money: ${remaining_money:.2f}")
print(f"Purchase price: ${purchase_price:.2f}")

# Make decisions
if remaining_money <= 0:
    print("\nRecommendation:")
    print("Your expenses are greater than or equal to your income.")
    print("Focus on reducing expenses before making purchases.")
else:
    percentage_used = (purchase_price / remaining_money) * 100

    print(f"Percentage of remaining money used: {percentage_used:.1f}%")

    if purchase_price > remaining_money:
        message = "You cannot afford this purchase."
    elif percentage_used > 50:
        message = "Think carefully before buying. This is a large expense."
    elif 20 <= percentage_used <= 50:
        message = "This purchase is reasonable, but make sure it fits your budget."
    else:
        message = "This purchase fits comfortably within your budget."

    print("\nRecommendation:")
    print(message)
````
