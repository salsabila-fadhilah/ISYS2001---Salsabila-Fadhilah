## Part 3: Building the Budget Tracker

Now for the real thing. Ask AI to help you write a program that:
- Takes the user's monthly income
- Takes up to 3 expenses
- Calculates what's left over
- Warns the user if the budget goes negative

> 💡 **Pro tip:** The more specific your prompt, the less clean-up you'll need to do afterwards. "Write a budget tracker" gets you something generic; describing the exact inputs, outputs, and edge cases gets you something closer to what you actually need.


```python
# Write your detailed prompt here:
"""
My Budget Tracker Prompt:

Write a Python program that acts as a simple monthly budget tracker.
The program should:
- Ask the user for their monthly income.
- Ask the user to enter up to three monthly expenses (rent, food, entertainment, or any expenses they choose).
- Calculate the total expenses.
- Calculate the remaining budget by subtracting the total expenses from the monthly income.
- Display the total expenses and the remaining budget.
- If the remaining budget is less than zero, display a warning message that the user is over budget.
- Otherwise, display a positive message encouraging the user to keep saving.
- Use float values for all money amounts and format all currency to two decimal places.
"""

# Budget Tracker Program

# Ask for monthly income
income = float(input("Enter your monthly income: $"))

# Ask for three expenses
expense1 = float(input("Enter your first expense: $"))
expense2 = float(input("Enter your second expense: $"))
expense3 = float(input("Enter your third expense: $"))

# Calculate total expenses
total_expenses = expense1 + expense2 + expense3

# Calculate remaining budget
remaining_budget = income - total_expenses

# Display results
print("\n----- Monthly Budget Summary -----")
print(f"Monthly Income: ${income:.2f}")
print(f"Total Expenses: ${total_expenses:.2f}")
print(f"Remaining Budget: ${remaining_budget:.2f}")

# Check if user is over budget
if remaining_budget < 0:
    print("⚠️ Warning: You are over budget! Consider reducing your expenses.")
else:
    print("✅ Great job! You are within your budget. Keep up the good work!")
```
