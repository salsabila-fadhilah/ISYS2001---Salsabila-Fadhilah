````python
# STEP 4: Your pseudocode
# Pseudocode is your plan in plain language. It does not need to run.

# 1. Start the program.
# 2. Ask the user for their name.
# 3. Ask the user for their favourite colour.
# 4. Ask the user for their favourite food, favourite movie, and favourite hobby.
# 5. Ask the user for their birth year.
# 6. Calculate the user's age using the current year.
# 7. Print a personalised summary message with all the information.
# 8. End the program.
````

## Step 5: Write the Code

**What to do:** Now turn your pseudocode into real Python, using only `input()`, `print()`, variables, and basic operations. Below is one example solution. Run it, then compare it line by line with the pseudocode you wrote.

````python
import datetime

# Collect user input for various preferences
name = input("What is your name? ")
favourite_colour = input("What's your favourite colour? ")
favourite_food = input("What's your favourite food? ")
favourite_movie = input("What's your favourite movie? ")
favourite_hobby = input("What's your favourite hobby? ")
birth_year = int(input("What year were you born? "))

# Calculate age
current_year = datetime.datetime.now().year
age = current_year - birth_year

# Construct and display a personalised summary message
print("\nHello, " + name + "!")
print("Your favourite colour is " + favourite_colour +
      ", you love " + favourite_food +
      ", your favourite movie is " + favourite_movie +
      ", and you enjoy " + favourite_hobby + ".")
print("You are", age, "years old!")
````

### Challenge 2: Age calculator

**What to do:** Ask the user for their birth year, work out their age for the current year, and print it.

> **Hint:** `input()` returns text, so convert the birth year with `int()` before doing the maths. You may use the `datetime` module to get the current year. If it is new to you, ask an AI tutor for a simple step-by-step explanation.

````python
import datetime

# Ask the user for their birth year
birth_year = int(input("What year were you born? "))

# Get the current year
current_year = datetime.datetime.now().year

# Calculate the user's age
age = current_year - birth_year

# Display the result
print("You are", age, "years old!")
````
